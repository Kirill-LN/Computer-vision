```python
from skimage import io
import cv2 as cv


# адреса картинок
image_path1 = 'https://wikiway.com/upload/hl-photo/fa3/fa3f6f31c695fb1cdeea2d33f24a99af/kinkakuji_14.jpg'
image_path2 = 'https://wikiway.com/upload/hl-photo/2fb/kinkakuji_13.jpg'

# Чтение, преобразование в нуужный цвет и вывод в отдельном окне
urls = [image_path1, image_path2]

def show_image(urls: dict):
    """ Функция чтения и вывода каринок
    Принимает на вход список из url адресов картинок, обрабатывает через cvtColor в нужный набор цветов
    Выводит коректное отображение картинок в правильном цвете.
    """

    for url in urls:
        image = io.imread(url)
        imageTrue = cv.cvtColor(image, cv.COLOR_RGB2BGR)
        cv.imshow('Image Kioto', imageTrue)
        cv.waitKey(0) # ждет нажатие любой кнопки для выхода
        cv.destroyAllWindows() # закрывает все окна
        

show_image(urls)
```

```python
import matplotlib.pyplot as plt


# вывод в самом jupyter/kolab с координатными осями
def show_in_jupyter(urls: dict):
    """Функция принимает на вход список с url изображения
    Выводит в ячейке jupyter/kolab выброный диапозон на изображении
    """
    for url in urls:
        image = io.imread(url)
        fig = plt.figure(figsize=(6,4))
        img = fig.add_subplot(1, 1, 1)
        img.imshow(image[100:250, 400:600])
        plt.show()


    
show_in_jupyter(urls)
```

```python
import matplotlib.pyplot as plt

# изменение цвета
def landscaping(urls: dict):
    """ Функция принимает на вход список с url адресами картинок
    Читает, меняет значение интенсивности 2 канала на пикселях которые соответсвуют условию в маске
    Получаем озеленение
    """
    
    for url in urls:
        image = io.imread(url)
        # Прописываем в максе условия интенсивности для поиска пикселей: ищем пиксели крансого больше 110, зеленого меньше 100 и синиго меньше 100
        mask = (
            (image[:, :, 0] > 110) &  # R — высокий
            (image[:, :, 1] < 100) &  # G — низкий
            (image[:, :, 2] < 100)    # B — низкий
        )

        # Меняем красный цвет на зелёный, озеленение))
        image[mask] = [0, 110, 0] 
        fig = plt.figure(figsize=(6,4))
        img = fig.add_subplot(1, 1, 1)
        img.imshow(image)
        plt.show()
    
landscaping(urls)
```
